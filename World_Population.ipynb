```python
dt.nlargest(10, 'pop2025').plot(x='country', y='pop2025', kind='bar', figsize=(8,5), title='Top 10 Populous Countries 2025')
plt.tight_layout(); plt.show()
```

```python
dt['density'].plot(kind='hist', bins=30, figsize=(6,4), title='Density Distribution')
plt.xlabel('Density'); plt.tight_layout(); plt.show()
```

```python
dt.plot.scatter(x='density', y='growthRate', alpha=0.6, figsize=(6,5), title='Growth vs Density')
plt.tight_layout(); plt.show()
```

```python
dt.nlargest(5, 'worldPercentage').set_index('country')['worldPercentage'].plot.pie(autopct='%1.1f%%')
plt.title('Top 5 Countries by World Share'); plt.ylabel(''); plt.tight_layout(); plt.show()
```

```python
dt_top = dt.nlargest(10, 'pop2025')[['country', 'pop2025', 'pop2050']]
dt_top.set_index('country').plot(kind='bar', figsize=(8,5), title='Population Growth (2025 vs 2050)')
plt.tight_layout(); plt.show()
```

```python
dt['density'].plot(kind='kde', title='Density KDE Curve')
plt.tight_layout(); plt.show()
```

```python
dt.plot.scatter(x='landAreaKm', y='pop2025', alpha=0.6, title='Population vs Land Area')
plt.tight_layout(); plt.show()
```

```python
dt.plot.scatter(x='pop2025', y='pop2050', title='Pop 2025 vs Pop 2050')
plt.tight_layout(); plt.show()
```
